# Caso práctico

### Se quiere estudiar la relación entre horas de estudio $x$ y la calificación obtenida en un examen $y$ tomando en cuenta los isguientes datos:

|Estudio (x)|Calificación (y)|
|---|---|
|2|4|
|3|5|
|5|7|
|7|10|
|9|12|


### 1. Planteamiento del modelo

El modelo de regresión lineal tiene la forma:

$y = mx + b$

Donde:

- $y$ es la variable dependiente (calificación)

- $x$ es la variable independiente (horas de estudio)

- $m$ es la pendiente de la recta (coeficiente de regresión)

- $b$ es la intersección de la recta (valor de calificación cuando no hay estudio)

El objetivo es encontrar m y b que minimicen la suma de los cuadrados de los errores (diferencias ente los valores observados $y_i$ y los valores predichos $mx_i + b$)

### 2. Ecuaciones de regresión lineal

# $m=\frac{n∑xy - (∑x)(∑y)}{n∑x^2 - (∑x)^2}$

# $b = \frac{∑y - m∑x}{n}$

### 3. Gráfica de los datos

```python
x = np.array([2, 3, 5, 7, 9])
y = np.array([4, 5, 7, 10, 12])

plt.scatter(x, y)

plt.xlabel('Estudio (x)')
plt.ylabel('Calificación (y)')

plt.show()
```

### 4. Cálculo

```python
n = 5 # Número de puntos
x_sum = np.sum(x) # Suma de los valores del vector x
y_sum = np.sum(y) # Suma de los valores del vector y
x_sum_sq = np.dot(x, x.T) # Suma de los cuadrados de x usando el producto punto
xy_sum = np.dot(x,y.T) # Suma del producto del vector x con y

print(x_sum)
print(y_sum)
print(x_sum_sq)
print(xy_sum)
```

```python
# Cálculo de m
m = ((n*xy_sum) - (x_sum*y_sum))/((n*x_sum_sq) - (x_sum**2))
print(m)
```

```python
# Cálulo de b
b = (y_sum - m*x_sum)/(n)
print(b)
```

### 5. Evaluación

```python
y_pred = np.array([m*i+b for i in x])
y_pred
```

```python
# Gráfica
x = np.array([2, 3, 5, 7, 9])
y = np.array([4, 5, 7, 10, 12])

plt.scatter(x, y)
plt.plot(x, y_pred, color='r')

plt.xlabel('Estudio (x)')
plt.ylabel('Calificación (y)')

plt.show()
```

### 6. Coeficiente de Determinación

# $R^2 = 1 - \frac{\sum(y_i - ypred_i)^2}{\sum(y_i - yprom)^2}$

Donde:
- $y_i$ son los valores predichos
- $yprom$ es la media de los valores observados

```python
# media
y_prom = np.mean(y)
print(y_prom)
```

```python
# Suma de cuadrados de errores

SSE = np.sum((y - y_pred)**2)
print(SSE)

SST = np.sum((y - y_prom)**2)
print(SST)

R_squared = 1 - (SSE/SST)
print(R_squared)
```
