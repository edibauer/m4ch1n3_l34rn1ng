# Metodo de mínimos cuadrados
Es una técnica utilizada para encontrar la recta (o curva), que mejor se ajusta aun conjunto de datos, minimizando la suma de los cuadrados de las diferencias entre los valores observados y los valores predichos del modelo

## Ejemplo
Se tienen los siguientes datos:

x|y
---|---
1|2
2|3
3|5
4|6

Se requiere encontrar la recta $y=mx + b$ que mejor se ajsute a esos datos

### 1. Planteamiento
La recta de mínimos cuadrados tiene la forma:

$y = mx + b$

Donde:
- $m$ es la pendiente
- $b$ es la ordenada al origen (intersección con $y$)

El objetivo es encontrar $m$ y $b$ que minimicen la suma de los cuadrados de los errores (diferencias entre los valores observados $y_i$ y los valores predichos $mx_i + b$)


### 2. Formula
Las formulas para calcular $m$ y $b$ son:

$m=\frac{n∑xy - (∑x)(∑y)}{n∑x^2 - (∑x)^2}$

$b = \frac{∑y - m∑x}{n}$

### 3. Gráfica de los datos

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 3, 5, 6])

plt.scatter(x,y)

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

### 4. Cálculo

```python
n = 4 # npumero de puntos
x_sum = np.sum(x) # Suma de los valores de x
y_sum = np.sum(y) # suma de los valores de y
x_sum_sq = np.sum(np.square(x)) # suma de los cuadrados de x
x_sum_sq2 = np.dot(x, x) # Suma de los cuadrados de x usando el producto punto
xy_sum = np.dot(x,y) # Suma de los producto de xy
print(x_sum)
print(y_sum)
print(x_sum_sq)
print(x_sum_sq2)
print(xy_sum)
```

```python
m = (n * xy_sum - (x_sum*y_sum))/(n*x_sum_sq - (x_sum)**2)
print(m)
```

```python
b = (y_sum - m*x_sum)/(n)
print(b)
```

### 5. Modelo
$y=1.4x+0.5$

### 6. Evaluación

```python
y_pred = np.array([m*i+b for i in x])
y_pred
```

```python
### 7. Gráfica
plt.scatter(x,y)
plt.plot(x, y_pred, color="r")

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
